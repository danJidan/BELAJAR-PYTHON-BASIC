## 1. Cara Membuat & Mengambil Isi List

Setiap barang di dalam list punya nomor urut indeks **mulai dari 0**.

```python
tas_sekolah = ["Buku Tulis", "Kotak Pensil", "Penggaris", "Botol Minum"]

print("Isi seluruh tas :", tas_sekolah)
print("Barang ke-0     :", tas_sekolah[0])   # Buku Tulis
print("Barang ke-1     :", tas_sekolah[1])   # Kotak Pensil
print("Barang terakhir :", tas_sekolah[-1])  # Botol Minum

# Mengganti isi barang
tas_sekolah[0] = "Buku Gambar"  # Ganti Buku Tulis jadi Buku Gambar
print("\nSetelah Buku Tulis diganti:", tas_sekolah)
```

---
## 2. Menambah Barang ke List

- **`.append(barang)`**: Menambahkan barang baru di **urutan paling belakang**.
- **`.insert(nomor, barang)`**: Menyelipkan barang baru di **nomor urut tertentu**.

```python
buah = ["Apel", "Jeruk", "Pisang"]
print("Awal           :", buah)

# 1. .append() -> Tambah di akhir
buah.append("Mangga")
print("Setelah append :", buah)

# 2. .insert() -> Selipkan di nomor urut 1
buah.insert(1, "Anggur")
print("Setelah insert :", buah)
```

---
## 3. Menghapus Barang dari List

- **`.remove(nama_barang)`**: Menghapus barang berdasarkan namanya.
- **`.pop()`**: Mengeluarkan dan mengambil barang paling belakang.

```python
makanan = ["Bakso", "Mie Ayam", "Siomay", "Batagor"]
print("Menu Awal      :", makanan)

# 1. .remove() menghapus Siomay
makanan.remove("Siomay")
print("Setelah remove :", makanan)

# 2. .pop() mengeluarkan makanan paling akhir (Batagor)
makanan_keluar = makanan.pop()
print(f"Setelah pop    : {makanan} (Makanan yang diambil: {makanan_keluar})")
```

---
## 4. Mengurutkan & Menghitung List

- **`len(list)`**: Menghitung ada berapa total barang di dalam list.
- **`.sort()`**: Mengurutkan nilai dari yang terkecil ke terbesar (atau A sampai Z).
- **`.reverse()`**: Membalikkan urutan isi list.

```python
nilai_ulangan = [80, 95, 70, 85, 90]

print("Daftar Nilai       :", nilai_ulangan)
print("Banyak Nilai (len) :", len(nilai_ulangan))

# Mengurutkan nilai dari kecil ke besar
nilai_ulangan.sort()
print("Setelah di-sort    :", nilai_ulangan)

# Membalik urutan
nilai_ulangan.reverse()
print("Setelah di-reverse :", nilai_ulangan)
```

## 5. 🎯 Latihan Seru: Manajemen Tas Game

Ayo buat simulasi tas inventori game petualangan!

```python
# Tas Awal Pemain
inventori = ["Pedang Kayu", "Ramuan Darah", "Roti"]
print("Tas Awal           :", inventori)

# 1. Menemukan Harta Karun 'Perisai Besi'
inventori.append("Perisai Besi")
print("Dapat Harta        :", inventori)

# 2. Memakan Roti saat lapar
inventori.remove("Roti")
print("Setelah makan Roti :", inventori)

print(f"\nJumlah barang di tas sekarang ada {len(inventori)} item.")
```
