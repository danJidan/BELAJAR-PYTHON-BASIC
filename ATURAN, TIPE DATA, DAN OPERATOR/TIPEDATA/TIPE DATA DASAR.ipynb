## 1. Apa itu Variabel?

> 📦 **Analogi:** Bayangkan kamu punya kotak kardus. Di luar kotak, kamu tempel stiker nama: `skor`. Di dalam kotak, kamu masukkan angka: `100`.

Di Python, cara membuatnya sangat mudah:
`nama_kotak = isinya`

```python
nama_siswa = "Rian"
umur = 14
kelas = 8

print("Nama Siswa :", nama_siswa)
print("Umur       :", umur, "tahun")
print("Kelas      :", kelas)
```

## 2. 4 Tipe Data Utama di Python

Ada 4 jenis data yang paling sering kita gunakan sehari-hari:

| Tipe Data | Nama Keren | Keterangan | Contoh |
| :--- | :---: | :--- | :--- |
| **Teks** | `str` (String) | Kata / Kalimat (pakai tanda petik) | `"Budi"`, `"SMP Negeri 1"` |
| **Bilangan Bulat** | `int` (Integer) | Angka tanpa koma | `14`, `100`, `-5` |
| **Bilangan Desimal** | `float` | Angka berkoma (pakai titik `.`) | `3.14`, `155.5`, `87.5` |
| **Benar / Salah** | `bool` (Boolean) | Nilai logika kebenaran | `True` (Benar) / `False` (Salah) |

```python
# 1. String (Teks)
nama_game = "Roblox"

# 2. Integer (Bilangan Bulat)
jumlah_koin = 250

# 3. Float (Bilangan Desimal)
tinggi_badan = 158.5  # cm

# 4. Boolean (True / False)
sudah_makan = True

print("Game Favorit :", nama_game)
print("Jumlah Koin  :", jumlah_koin)
print("Tinggi Badan :", tinggi_badan, "cm")
print("Sudah Makan? :", sudah_makan)
```

## 3. Cara Cek Tipe Data (`type()`)

Jika kamu penasaran apa jenis tipe data suatu variabel, gunakan perintah ajaib `type()`.

```python
nama = "Aisyah"
umur = 13
nilai_ipa = 92.5
apakah_juara = True

print("Tipe data nama        :", type(nama))
print("Tipe data umur        :", type(umur))
print("Tipe data nilai_ipa   :", type(nilai_ipa))
print("Tipe data apakah_juara:", type(apakah_juara))
```

## 4. Mengubah Tipe Data (Type Casting Sederhana)

Terkadang kita ingin mengubah angka dalam teks menjadi angka beneran agar bisa dihitung matematika.

- `int("10")` $\rightarrow$ Mengubah teks `"10"` menjadi angka bulat `10`.
- `float("7.5")` $\rightarrow$ Mengubah teks `"7.5"` menjadi angka desimal `7.5`.
- `str(100)` $\rightarrow$ Mengubah angka `100` menjadi teks `"100"`.

```python
# Masalah jika masih berupa teks (String Concatenation):
angka_teks_1 = "10"
angka_teks_2 = "20"
hasil_teks = angka_teks_1 + angka_teks_2
print("Jika belum diubah tipe datanya :", hasil_teks)  # Menghasilkan "1020"

# Solusi: Ubah ke integer dulu menggunakan int()
angka_asli_1 = int(angka_teks_1)
angka_asli_2 = int(angka_teks_2)
hasil_hitung = angka_asli_1 + angka_asli_2
print("Setelah diubah ke int (dihitung):", hasil_hitung) # Menghasilkan 30
```

## 5. 🎯 Latihan Seru: Biodata Karakter Game

Ayo buat data karakter jagoanmu dengan menggabungkan semua tipe data yang sudah dipelajari!

```python
# Profil Karakter Game
nama_hero = "Satria Petir"
level = 15
darah = 100.0
punya_pedang = True

print("=" * 30)
print("     PROFIL HERO GAME         ")
print("=" * 30)
print(f"Nama Hero    : {nama_hero}")
print(f"Level        : {level}")
print(f"Sisa Darah   : {darah} %")
print(f"Punya Senjata: {punya_pedang}")
print("=" * 30)
```
