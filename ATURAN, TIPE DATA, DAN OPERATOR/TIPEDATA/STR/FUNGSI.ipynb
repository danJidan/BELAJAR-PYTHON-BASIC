## 1. Mengambil Huruf (Indexing & Slicing)

Setiap huruf di dalam teks memiliki nomor antrean (indeks), **dimulai dari angka 0** dari kiri.
- `kata[0]` = Huruf pertama
- `kata[-1]` = Huruf paling terakhir
- `kata[0:3]` = Mengambil potongan dari huruf ke-0 sampai ke-2 (huruf ke-3 tidak ikut).

```python
kata = "PYTHON"

print("Kata asli            :", kata)
print("Huruf pertama [0]    :", kata[0])   # P
print("Huruf kedua [1]      :", kata[1])   # Y
print("Huruf terakhir [-1]  :", kata[-1])  # N
print("Potongan 3 huruf [:3]:", kata[:3])  # PYT
```

---
## 2. Mengubah Huruf Besar & Kecil

Python punya tombol otomatis untuk mengatur ukuran huruf:
- **`.upper()`**: Mengubah semua huruf jadi **KAPITAL** (seperti sedang berteriak).
- **`.lower()`**: Mengubah semua huruf jadi **kecil** semua.
- **`.title()`**: Mengubah huruf **Awal Setiap Kata Jadi Kapital** (cocok untuk nama orang).

```python
nama_lengkap = "muhammad jidan santoso"

print("Asli     :", nama_lengkap)
print(".upper() :", nama_lengkap.upper())  # MUHAMMAD JIDAN SANTOSO
print(".lower() :", nama_lengkap.lower())  # muhammad jidan santoso
print(".title() :", nama_lengkap.title())  # Muhammad Jidan Santoso
```

---
## 3. Menghitung & Mencari Kata

- **`len(teks)`**: Menghitung berapa jumlah huruf/karakter.
- **`.count("kata")`**: Menghitung berapa kali suatu kata muncul.
- **`.find("kata")`**: Mencari tahu di urutan ke berapa kata itu berada.

```python
kalimat = "Saya suka belajar Python karena Python itu seru"

print(f"Panjang kalimat: {len(kalimat)} karakter")
print("Jumlah kata 'Python' muncul :", kalimat.count("Python"))
print("Jumlah huruf 'a' muncul     :", kalimat.count("a"))
print("Kata 'belajar' ada di posisi :", kalimat.find("belajar"))
```

---
## 4. Membersihkan & Mengganti Teks

- **`.strip()`**: Menghapus spasi tidak sengaja di awal dan akhir teks.
- **`.replace(kata_lama, kata_baru)`**: Mengganti kata lama dengan kata baru.

```python
# 1. .strip() membersihkan spasi berlebih
username = "   jidan_gamer   "
print("Sebelum strip :", repr(username))
print("Sesudah strip :", repr(username.strip()))

# 2. .replace() mengganti kata
status = "Saya benci hari Senin"
status_positif = status.replace("benci", "suka")
print("\nStatus Asli  :", status)
print("Status Baru  :", status_positif)
```

---
## 5. Memecah & Menggabungkan Teks

- **`.split()`**: Memecah satu kalimat panjang menjadi daftar kata-kata terpisah.
- **`.join()`**: Menggabungkan kembali kumpulan kata menjadi satu kalimat utuh.

```python
# 1. .split() memecah kalimat
daftar_hobi = "Membaca,Gaming,Sepakbola,Coding"
hobi_list = daftar_hobi.split(",")
print("Hasil Split menjadi List:", hobi_list)

# 2. .join() menyatukan kata dengan pemisah tanda hubung
kata_pecah = ["Belajar", "Python", "Itu", "Mudah"]
kalimat_gabung = " - ".join(kata_pecah)
print("Hasil Join dengan ' - ' :", kalimat_gabung)
```

## 6. 🎯 Latihan Seru: Sensor Kata Rahasia

Bayangkan kamu membuat program chat game yang otomatis menyensor kata buruk!

```python
chat_masuk = "Kamu mainnya jelek banget deh!"

# Sensor kata 'jelek' menjadi '***'
chat_bersih = chat_masuk.replace("jelek", "***")

print("Chat Asli   :", chat_masuk)
print("Chat Sensor :", chat_bersih)
```
