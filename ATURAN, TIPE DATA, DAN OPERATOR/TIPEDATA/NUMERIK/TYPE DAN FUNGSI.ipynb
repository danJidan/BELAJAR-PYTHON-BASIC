## 1. Mengenal 2 Jenis Angka: Integer & Float

1. **`int` (Integer):** Bilangan bulat (tidak ada komanya). Contoh: jumlah siswa `32`, umur `14`.
2. **`float` (Float):** Bilangan desimal / pecahan berkoma (menggunakan tanda titik `.`). Contoh: nilai `88.5`, tinggi `160.2`.

```python
jumlah_buku = 12       # Integer (bilangan bulat)
harga_pensil = 2500.50 # Float (bilangan desimal)

print("Jumlah Buku  :", jumlah_buku, "| Tipe:", type(jumlah_buku))
print("Harga Pensil :", harga_pensil, "| Tipe:", type(harga_pensil))
```

## 2. Fungsi Hitung Bawaan Python

Python punya alat bantu hitung bawaan yang langsung bisa kita panggil:

| Fungsi | Kegunaan | Contoh | Hasil |
| :--- | :--- | :--- | :--- |
| **`abs(x)`** | Mengubah angka negatif jadi positif (nilai mutlak) | `abs(-10)` | `10` |
| **`round(x, n)`** | Membulatkan angka desimal ke `n` angka di belakang koma | `round(85.678, 2)` | `85.68` |
| **`min(...)`** | Mencari angka paling kecil / terendah | `min(80, 95, 70)` | `70` |
| **`max(...)`** | Mencari angka paling besar / tertinggi | `max(80, 95, 70)` | `95` |
| **`sum(list)`** | Menjumlahkan seluruh isi kumpulan angka | `sum([10, 20, 30])` | `60` |

```python
# 1. abs() -> Nilai mutlak (selalu positif)
jarak = abs(-50)
print("Jarak mutlak dari -50 adalah:", jarak)

# 2. round() -> Pembulatan nilai
nilai_asli = 87.647
print("Nilai asli        :", nilai_asli)
print("Bulat 1 angka koma:", round(nilai_asli, 1))
print("Bulat tanpa koma  :", round(nilai_asli))

# 3. min(), max(), dan sum()
daftar_skor = [75, 90, 65, 88, 95]
print("\nDaftar Skor :", daftar_skor)
print("Skor Tertinggi (max) :", max(daftar_skor))
print("Skor Terendah  (min) :", min(daftar_skor))
print("Total Seluruh Skor   :", sum(daftar_skor))
```

## 3. Bantuan Modul Matematika (`math`)

Untuk rumus matematika sekolah seperti akar kuadrat atau pembulatan khusus, kita bisa memanggil modul bawaan bernama `math` dengan mengetik `import math`.

```python
import math

# 1. Akar Kuadrat (math.sqrt)
angka = 64
print(f"Akar kuadrat dari {angka} adalah :", math.sqrt(angka))  # Hasil: 8.0

# 2. Pembulatan ke Atas (math.ceil) dan ke Bawah (math.floor)
nilai_desimal = 4.2
print("\nNilai awal :", nilai_desimal)
print("Bulat ke atas  (math.ceil) :", math.ceil(nilai_desimal))   # 5
print("Bulat ke bawah (math.floor):", math.floor(nilai_desimal))  # 4

# 3. Nilai Pi (math.pi) untuk Luas Lingkaran
jari_jari = 7
luas_lingkaran = math.pi * (jari_jari ** 2)
print(f"\nLuas lingkaran r={jari_jari} cm adalah : {luas_lingkaran:.2f} cm²")
```

## 4. 🎯 Latihan Seru: Menghitung Nilai Rapor

Mari kita buat program untuk menghitung total nilai dan rata-rata rapor siswa!

```python
# Nilai Mata Pelajaran SMP
nilai_matematika = 85
nilai_ipa = 90
nilai_bahasa_indonesia = 88
nilai_bahasa_inggris = 92

# Hitung Total dan Rata-rata
total_nilai = nilai_matematika + nilai_ipa + nilai_bahasa_indonesia + nilai_bahasa_inggris
jumlah_mapel = 4
rata_rata = total_nilai / jumlah_mapel

print("=" * 35)
print("       HASIL NILAI RAPOR       ")
print("=" * 35)
print(f"Total Nilai : {total_nilai}")
print(f"Rata-rata   : {rata_rata:.2f}")
print(f"Nilai Max   : {max(nilai_matematika, nilai_ipa, nilai_bahasa_indonesia, nilai_bahasa_inggris)}")
print("=" * 35)
```
