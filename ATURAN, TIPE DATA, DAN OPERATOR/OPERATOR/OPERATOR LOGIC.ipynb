## 1. Operator Perbandingan

Digunakan untuk membandingkan dua nilai:

| Simbol | Arti | Contoh | Hasil |
| :---: | :--- | :---: | :---: |
| **`==`** | Sama persis | `5 == 5` | `True` |
| **`!=`** | Tidak sama dengan | `5 != 3` | `True` |
| **`>`** | Lebih besar | `10 > 7` | `True` |
| **`<`** | Lebih kecil | `4 < 2` | `False` |
| **`>=`** | Lebih besar atau sama | `10 >= 10` | `True` |
| **`<=`** | Lebih kecil atau sama | `8 <= 5` | `False` |

```python
nilai_kkm = 75
nilai_siswa = 85

print("Apakah nilai siswa sama dengan KKM? (==) :", nilai_siswa == nilai_kkm)
print("Apakah nilai siswa di atas KKM?      (>)  :", nilai_siswa > nilai_kkm)
print("Apakah nilai siswa di bawah KKM?    (<)  :", nilai_siswa < nilai_kkm)
```

## 2. 3 Operator Logika Utama (`and`, `or`, `not`)

1. **`and` (DAN):** Bernilai `True` **hanya jika SEMUA syarat terpenuhi**.
   - *Contoh:* Untuk upacara bendera harus bawa topi `and` pakai dasi.
2. **`or` (ATAU):** Bernilai `True` jika **MINIMAL SALAH SATU syarat terpenuhi**.
   - *Contoh:* Beli jajan boleh bayar tunai `or` pakai QRIS.
3. **`not` (BUKAN / KEBALIKAN):** Membalik nilai (`True` jadi `False`, `False` jadi `True`).

```python
# 1. Contoh 'and' (Semua syarat wajib True)
punya_tiket = True
tinggi_cukup = True
boleh_masuk_wahana = punya_tiket and tinggi_cukup
print("Boleh Masuk Wahana (and) :", boleh_masuk_wahana)

# 2. Contoh 'or' (Salah satu True sudah cukup)
punya_kartu_member = False
punya_kupon_diskon = True
dapat_potongan_harga = punya_kartu_member or punya_kupon_diskon
print("Dapat Diskon (or)        :", dapat_potongan_harga)

# 3. Contoh 'not' (Membalik nilai)
hujan = False
print("Bisa main bola? (not hujan):", not hujan)
```

## 3. Mengecek Anggota (`in` & `not in`)

Sangat berguna untuk mencari apakah namamu ada di dalam daftar absen atau kelompok!

```python
tim_futsal = ["Andi", "Budi", "Candra", "Dedi"]

print("Apakah 'Budi' ada di tim?  :", "Budi" in tim_futsal)     # True
print("Apakah 'Zaki' ada di tim?  :", "Zaki" in tim_futsal)     # False
print("Apakah 'Zaki' BUKAN di tim?:", "Zaki" not in tim_futsal) # True
```

## 4. 🎯 Studi Kasus: Syarat Naik Roller Coaster

**Aturan Taman Bermain:**
Pengunjung boleh naik wahana jika **Tinggi Badan minimal 140 cm** `DAN` **Umur minimal 12 tahun**.

```python
# Data Pengunjung
nama_pengunjung = "Fajar"
tinggi_badan = 145  # cm
umur = 13          # tahun

# Evaluasi Syarat Logika
syarat_tinggi = tinggi_badan >= 140
syarat_umur = umur >= 12
boleh_naik = syarat_tinggi and syarat_umur

print("=" * 45)
print("     PEMERIKSAAN TIKET ROLLER COASTER     ")
print("=" * 45)
print(f"Nama Pengunjung : {nama_pengunjung}")
print(f"Tinggi Badan    : {tinggi_badan} cm (Syarat Tinggi: {syarat_tinggi})")
print(f"Umur            : {umur} tahun (Syarat Umur: {syarat_umur})")
print("-" * 45)
print(f"STATUS          : {'SELAMAT MENIKMATI WAHANA! 🎢' if boleh_naik else 'MAAF, BELUM MEMENUHI SYARAT 🚫'}")
print("=" * 45)
```
