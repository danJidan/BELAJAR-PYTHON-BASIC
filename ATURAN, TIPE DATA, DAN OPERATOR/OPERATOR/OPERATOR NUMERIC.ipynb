# ➕ OPERATOR HITUNG MATEMATIKA (NUMERIK) DI PYTHON

Halo teman-teman! 👋 Di Python, kita bisa melakukan berbagai macam operasi matematika seperti di kalkulator dengan simbol-simbol operator yang mudah diingat.

---

## 📋 Daftar Simbol Hitung:
1. **`+`** (Tambah)
2. **`-`** (Kurang)
3. **`*`** (Kali)
4. **`/`** (Bagi Desimal)
5. **`//`** (Bagi Bulat / Sisa Dibuang)
6. **`%`** (Modulus / Sisa Hasil Bagi)
7. **`**`** (Pangkat)
8. **`+=` dan `-=`** (Singkatan Tambah & Kurang)

---
## 1. 7 Simbol Hitung Utama

```python
a = 10
b = 3

print(f"a = {a}, b = {b}")
print("-" * 30)
print("1. Penjumlahan (10 + 3)  =", a + b)   # 13
print("2. Pengurangan (10 - 3)  =", a - b)   # 7
print("3. Perkalian   (10 * 3)  =", a * b)   # 30
print("4. Pembagian   (10 / 3)  =", a / b)   # 3.3333...
print("5. Bagi Bulat  (10 // 3) =", a // b)  # 3 (koma dibuang)
print("6. Sisa Bagi   (10 % 3)  =", a % b)   # 1 (karena 10 dibagi 3 dapat 3 sisa 1)
print("7. Pangkat     (10 ** 3) =", a ** b)  # 1000 (10 x 10 x 10)
```

## 2. Mengenal Modulus (`%`) untuk Ganjil / Genap

Operator `%` mencari **sisa pembagian**.
- Jika sebuah angka dibagi 2 sisanya `0`, berarti angka itu **GENAP**.
- Jika sebuah angka dibagi 2 sisanya `1`, berarti angka itu **GANJIL**.

```python
angka1 = 8
angka2 = 9

print(f"8 % 2 = {angka1 % 2} -> Angka 8 adalah GENAP")
print(f"9 % 2 = {angka2 % 2} -> Angka 9 adalah GANJIL")
```

## 3. Singkatan Hitung (`+=` dan `-=`)

Daripada menulis `skor = skor + 10`, kita bisa menulis singkat: `skor += 10`.
Sangat sering dipakai untuk menambah skor atau mengurangi darah di game!

```python
skor_game = 100
print("Skor awal          :", skor_game)

skor_game += 50   # Skor bertambah 50
print("Dapat koin (+50)   :", skor_game)

skor_game -= 30   # Terkena duri (-30)
print("Kena jebakan (-30) :", skor_game)
```

## 4. 🎯 Latihan Seru: Total Belanja di Kantin Sekolah

Mari kita hitung total uang jajan dan kembalian di kantin!

```python
harga_es_teh = 3000
jumlah_es_teh = 2

harga_gorengan = 1500
jumlah_gorengan = 4

# Hitung Subtotal
total_es_teh = harga_es_teh * jumlah_es_teh
total_gorengan = harga_gorengan * jumlah_gorengan
total_bayar = total_es_teh + total_gorengan

uang_dibayar = 20000
kembalian = uang_dibayar - total_bayar

print("=" * 35)
print("       STRUK KANTIN SMP GG       ")
print("=" * 35)
print(f"Es Teh ({jumlah_es_teh}x)    : Rp {total_es_teh:,}")
print(f"Gorengan ({jumlah_gorengan}x)  : Rp {total_gorengan:,}")
print("-" * 35)
print(f"Total Belanja : Rp {total_bayar:,}")
print(f"Uang Tunai    : Rp {uang_dibayar:,}")
print(f"Kembalian     : Rp {kembalian:,}")
print("=" * 35)
```
