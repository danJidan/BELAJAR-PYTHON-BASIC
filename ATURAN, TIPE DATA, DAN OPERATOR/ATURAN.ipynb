## 1. Aturan 1: Huruf Besar & Huruf Kecil itu Beda! (Case Sensitive)

Python sangat teliti dengan huruf besar (kapital) dan huruf kecil.
- Kata `skor`, `Skor`, dan `SKOR` dianggap **3 hal yang berbeda** oleh komputer!
- Perintah bawaan Python seperti `print()` harus ditulis huruf kecil semua. Jika kamu menulis `Print()` dengan P besar, komputer akan bingung dan memunculkan pesan error.

```python
# Contoh: 3 kotak nama yang berbeda
nama = "Budi"        # semua huruf kecil
Nama = "Siti"        # N huruf besar
NAMA = "Dan Jidan"   # semua huruf besar

print("Isi nama :", nama)
print("Isi Nama :", Nama)
print("Isi NAMA :", NAMA)

# Perhatikan: Ketiganya tidak saling menimpa karena dianggap berbeda!
```

```python
# ⚠️ Contoh jika kita salah mengetik huruf besar pada perintah bawaan:
# Print("Halo teman-teman")  # ❌ Ini akan error karena P besar

# ✅ Yang benar:
print("Halo teman-teman! Saya belajar Python.")
```

## 2. Aturan 2: Cara Memberi Nama Variabel

Variabel itu seperti **kotak mainan yang kita beri label nama**. Saat memberi nama kotak, ada aturannya:

### ✅ Boleh & Benar:
- Menggunakan huruf (`a-z`, `A-Z`), angka (`0-9`), dan garis bawah (`_`).
- Contoh: `skor_game`, `pemain1`, `level_up`, `_darah`.

### ❌ Dilarang (Bikin Error):
1. **Jangan diawali angka:** `1pemain` ❌ (Gantilah jadi `pemain1` atau `pemain_1` ✅).
2. **Jangan pakai spasi:** `skor game` ❌ (Gunakan garis bawah: `skor_game` ✅).
3. **Jangan pakai simbol:** `skor-game`, `total$`, `nama@` ❌.

```python
# Contoh nama variabel yang keren dan benar:
nama_karakter = "Hero GG"
nyawa = 100
level_pemain_1 = 5

print(f"Karakter : {nama_karakter}")
print(f"Nyawa    : {nyawa}")
print(f"Level    : {level_pemain_1}")
```

## 3. Aturan 3: Indentasi (Kerapian Baris Spasi)

Di pelajaran Bahasa Indonesia, saat menulis paragraf baru kita sering memajukan tulisan sedikit ke kanan.
Di Python, menjorokkan tulisan ke kanan (biasanya **4 spasi**) disebut **Indentasi**.

- Indentasi digunakan untuk menunjukkan kalimat/perintah mana yang menjadi bagian dari suatu kondisi (seperti `if`).

```python
nilai_ulangan = 85

if nilai_ulangan >= 75:
    # Baris ini menjorok 4 spasi ke kanan (masuk ke dalam kondisi if)
    print("Hebat! Kamu lulus ulangan!")
    print("Dapat hadiah es krim 🍦")
else:
    # Baris ini menjorok 4 spasi ke kanan (masuk ke dalam kondisi else)
    print("Jangan menyerah, yuk belajar lagi!")

# Baris ini kembali ke kiri (selalu dijalankan)
print("Program selesai.")
```

---
## 4. Aturan 4: Catatan Rahasia (Komentar `#`)

Komentar adalah catatan yang kita tulis di dalam kode. Komputer **tidak akan membaca atau menjalankan** komentar.
- Digunakan sebagai pengingat untuk diri sendiri atau teman kelompok.
- Ditulis menggunakan tanda pagar (**`#`**).

```python
# Ini adalah catatan: Program menghitung uang jajan

uang_saku = 20000  # Uang saku dari ibu
beli_bakso = 12000 # Beli bakso di kantin

# Hitung sisa uang
sisa_uang = uang_saku - beli_bakso

print("Sisa uang jajan saya: Rp", sisa_uang)
```

## 5. Aturan 5: Menulis Kalimat (Tanda Petik)

Jika ingin menulis kata atau kalimat (teks), kita harus membungkusnya dengan tanda petik:
- Boleh tanda petik satu: `'Halo'`
- Boleh tanda petik dua: `"Halo"`
- Jika di dalam kalimat ada tanda petik satu (seperti kata `Jum'at`), gunakan tanda petik dua di luarnya: `"Hari Jum'at"`.

```python
pesan1 = "Halo sahabat!"
pesan2 = 'Ayo belajar coding'
pesan3 = "Hari ini hari Jum'at yang cerah."

print(pesan1)
print(pesan2)
print(pesan3)
```

## 6. Aturan 6: Kata-Kata Khusus Python (Keywords)

Python sudah punya 'kata ajaib' bawaan yang sudah memiliki fungsi khusus, misalnya: `if`, `else`, `for`, `while`, `print`, `True`, `False`.
- Kita **tidak boleh** memakai kata-kata ini sebagai nama variabel.

```python
# ❌ if = 10       -> ERROR karena 'if' adalah kata kunci perintah
# ✅ nilai_if = 10 -> BENAR!

nilai_if = 10
print("Nama variabel nilai_if berhasil dibuat:", nilai_if)
```
