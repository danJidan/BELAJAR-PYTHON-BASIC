## 1. Persiapan Akun GitHub 🐙

GitHub adalah tempat para programmer di seluruh dunia menyimpan, membagikan, dan bekerja sama membuat aplikasi/proyek coding (seperti media sosial khusus coding!).

### Langkah Membuat Akun:
1. Siapkan **Akun Email** (Gmail / akun sekolah).
2. Buka website: [github.com](https://github.com/) lalu klik **Sign Up**.
3. Masukkan email, buat password yang aman, dan tentukan username kalian.
4. Lakukan verifikasi captcha & cek kode konfirmasi di email kalian. Selesai! 🎉

## 2. Menghubungkan Repository GitHub ke VS Code 🚀

Link Repository Belajar Python:
`https://github.com/danJidan/BELAJAR-PYTHON-BASIC.git`

### 📝 Perintah Git yang Sering Digunakan:

| Perintah (Command) | Fungsi / Kegunaan |
| :--- | :--- |
| `git clone <link-repo>` | Mendownload seluruh materi belajar dari GitHub ke laptop kalian |
| `git branch nama-kamu` | Membuat "cabang" catatan baru khusus milikmu sendiri |
| `git checkout nama-kamu` | Berpindah ke cabang catatan milikmu |
| `git add .` | Menandai semua perubahan file yang baru kalian edit |
| `git commit -m "pesan"` | Menyimpan catatan perubahan dengan pesan yang jelas |
| `git push origin nama-kamu` | Mengirim hasil belajarmu ke GitHub online |

## 3. Panduan Instalasi di Laptop / PC Sendiri 💻

### A. Pasang Python:
1. Buka website resmi: [python.org/downloads](https://www.python.org/downloads/).
2. Unduh versi Python terbaru untuk Windows.
3. Buka file installer yang terunduh.
4. ⚠️ **SANGAT PENTING (JANGAN LUPA):** Centang kotak **"Add python.exe to PATH"** di bagian paling bawah installer sebelum menekan tombol *Install Now*!
5. Klik **Install Now** dan tunggu hingga selesai.

### B. Pasang Visual Studio Code (VS Code):
1. Buka website: [code.visualstudio.com](https://code.visualstudio.com/).
2. Unduh untuk Windows dan pasang seperti aplikasi biasa (*Next > Next > Finish*).
3. Buka VS Code, klik ikon kotak-kotak di menu sebelah kiri (**Extensions** atau tekan `Ctrl + Shift + X`).
4. Cari dan klik **Install** pada:
   - **Python** (oleh Microsoft)
   - **Jupyter** (oleh Microsoft)
