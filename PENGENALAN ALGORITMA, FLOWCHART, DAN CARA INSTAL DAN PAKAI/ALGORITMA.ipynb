## 1. Apa itu Algoritma?

**Algoritma** adalah **urutan langkah-langkah logis dan sistematis untuk menyelesaikan suatu masalah atau mencapai tujuan tertentu.**

Komputer itu sebenarnya tidak bisa berpikir sendiri. Komputer adalah alat yang sangat cepat dan penurut, tetapi dia membutuhkan instruksi langkah-demi-langkah yang sangat jelas dari kita (manusia).

### 🤖 3 Ciri Algoritma yang Baik:
1. **Jelas & Pasti (Unambiguous):** Setiap langkah tidak membingungkan atau multi-tafsir.
2. **Berurutan (Sequential):** Langkah 1 dikerjakan sebelum langkah 2.
3. **Ada Awal & Akhir (Finite):** Masalah terselesaikan dan program berhenti (tidak berputar macet selamanya).

## 2. Pola Dasar Program: Input ➡️ Proses ➡️ Output

Hampir semua program komputer bekerja dengan pola sederhana:
1. **Input (Masukan):** Data yang dimasukkan (misal: panjang dan lebar lapangan).
2. **Proses (Pengolahan):** Perhitungan matematika atau tindakan (misal: panjang × lebar).
3. **Output (Keluaran):** Hasil yang ditampilkan ke layar (misal: luas lapangan).

```python
# Contoh Algoritma 1: Menghitung Luas Lapangan Futsal SMP
# 1. INPUT (Data masukan)
panjang = 25  # meter
lebar = 15    # meter

# 2. PROSES (Perhitungan matematika)
luas = panjang * lebar

# 3. OUTPUT (Menampilkan hasil)
print("=== HASIL PERHITUNGAN ALGORITMA ===")
print(f"Panjang lapangan : {panjang} meter")
print(f"Lebar lapangan   : {lebar} meter")
print(f"Luas lapangan    : {luas} meter persegi")
```

## 3. Algoritma dengan Pilihan / Kondisi

Dalam kehidupan sehari-hari, seringkali ada pilihan:
- *Jika hari hujan, bawa payung. Jika tidak hujan, tidak perlu bawa payung.*
- *Jika nilai ulangan >= 75, maka LULUS. Jika kurang, maka REMEDIAL.*

Di Python, kita bisa menggunakan perintah `if` dan `else` untuk membuat keputusan:

```python
# Contoh Algoritma 2: Menentukan Kelulusan Nilai Ulangan Informatika
nama_siswa = "Rian"
nilai_ulangan = 82

print(f"Nama Siswa : {nama_siswa}")
print(f"Nilai      : {nilai_ulangan}")

# Logika Algoritma (Pilihan/Keputusan)
if nilai_ulangan >= 75:
    print("Status     : SELAMAT, KAMU LULUS! 🎉")
else:
    print("Status     : Semangat, yuk ikut remedial dan belajar lagi! 💪")
```

## 4. 🎯 Tantangan Kreatif: Algoritma Kasir Kantin Sekolah

Coba jalankan kode di bawah ini! Kamu bisa mengubah angka `uang_saku`, `harga_es_teh`, atau `harga_roti` untuk melihat bagaimana algoritma menghitung sisa uang jajarmu.

```python
# Tantangan: Algoritma Kasir Kantin SMP
uang_saku = 20000
harga_es_teh = 4000
harga_roti = 5000

# 1. Hitung total belanja
total_belanja = harga_es_teh + harga_roti

# 2. Hitung sisa uang saku
sisa_uang = uang_saku - total_belanja

print("=== STRUK KASIR KANTIN SMP GG ===")
print("Uang Saku Anda   : Rp", uang_saku)
print("Total Belanja    : Rp", total_belanja)
print("Sisa Uang Saku   : Rp", sisa_uang)
print("-" * 35)

if sisa_uang >= 5000:
    print("Catatan: Asyik! Masih ada sisa untuk ditabung di celengan 🪙")
else:
    print("Catatan: Uang saku sudah pas-pasan, hemat ya! 😊")
```
