## 1. 📐 Mengenal Simbol-Simbol Standar Flowchart Beserta Gambarnya

Agar flowchart dapat dibaca dan dimengerti oleh programmer di seluruh penjuru dunia, bentuk simbolnya telah distandardisasi oleh **ANSI** (*American National Standards Institute*).

Setiap bentuk bangun datar memiliki arti dan tugas yang sangat spesifik:

```mermaid
graph TD
    classDef term fill:#4CAF50,stroke:#2E7D32,stroke-width:2px,color:#fff;
    classDef proc fill:#2196F3,stroke:#1565C0,stroke-width:2px,color:#fff;
    classDef io fill:#FF9800,stroke:#E65100,stroke-width:2px,color:#fff;
    classDef dec fill:#E91E63,stroke:#880E4F,stroke-width:2px,color:#fff;
    classDef prep fill:#9C27B0,stroke:#4A148C,stroke-width:2px,color:#fff;
    classDef conn fill:#00BCD4,stroke:#006064,stroke-width:2px,color:#fff;

    T1([1. Terminator: Mulai / Selesai]):::term --> P1[2. Process: Perhitungan / Aksi]:::proc
    P1 --> IO1[/3. Input / Output: Masukan Data / Tampilkan Hasil/]:::io
    IO1 --> D1{4. Decision: Pilihan Ya / Tidak?}:::dec
    D1 -- Ya --> PR1{{5. Preparation: Set Nilai Awal}}:::prep
    D1 -- Tidak --> C1((6. Connector)):::conn
```

---
### 📋 Tabel Rangkuman Simbol Flowchart:

| Gambar Visual | Nama Simbol | Nama Bentuk | Fungsi & Penjelasan | Contoh Isi Simbol |
| :---: | :--- | :--- | :--- | :--- |
| `([ Mulai ])` | **Terminator** | Oval / Kapsul | Menandai titik **AWAL (Start)** dan titik **AKHIR (End)** dari alur program. | `Mulai`, `Selesai`, `Start`, `Stop` |
| `[/ Input /]` | **Input / Output** | Jajar Genjang | Mengambil data masukan dari pengguna ATAU menampilkan hasil/output ke layar. | `Input Nilai Ujian`, `Tampilkan Luas` |
| `[ Proses ]` | **Process** | Persegi Panjang | Menjalankan tindakan, instruksi eksekusi, atau perhitungan rumus matematika. | `Luas = P × L`, `Hitung Total`, `Nilai = Nilai + 1` |
| `{ Pilihan? }`| **Decision** | Belah Ketupat (*Diamond*) | Menguji kondisi/keputusan logis. Memiliki **minimal 2 cabang keluar** (*Ya/True* dan *Tidak/False*). | `Apakah Nilai >= 75?`, `Apakah Saldo Cukup?` |
| `{{ Inisialisasi }}`| **Preparation** | Segi Enam (*Hexagon*) | Menyiapkan nilai awal variabel atau counter pengulangan sebelum proses dimulai. | `counter = 0`, `i = 1 to 10` |
| `(( A ))` | **On-Page Connector** | Lingkaran Kecil | Menyambungkan alur diagram pada **halaman yang sama** agar garis panah tidak saling silang ruwet. | `A`, `1` |
| `[\ B /]` | **Off-Page Connector** | Segi Lima (Rumah Terbalik) | Menyambungkan alur diagram ke **halaman/lembar kertas berikutnya**. | `Halaman 2` |
| `[[ Fungsi ]]` | **Predefined Process** | Persegi Panjang Garis Ganda | Menjalankan modul fungsi atau subprogram yang sudah didefinisikan terpisah. | `Panggil Fungsi CetakStruk()` |
| `───►` | **Flow Line** | Garis Panah | Menunjukkan **arah aliran logika** dari satu langkah ke langkah berikutnya. | Menghubungkan satu simbol ke simbol lain |

---
### 🌟 4 Aturan Emas Menggambar Flowchart:
1. **Arah Aliran:** Selalu mengalir dari **atas ke bawah** atau dari **kiri ke kanan**.
2. **Keteraturan Terminator:** Flowchart **HARUS** diawali tepat 1 simbol `Mulai` dan diakhiri dengan simbol `Selesai`.
3. **Kejelasan Percabangan (Decision):** Setiap garis panah yang keluar dari belah ketupat **wajib diberi label yang jelas** (misal: *Ya / Tidak*, atau *True / False*).
4. **Kalimat Singkat & Jelas:** Tuliskan instruksi di dalam simbol secara padat (gunakan kata kerja aktif seperti `Hitung`, `Tampilkan`, `Cek`).

## 2. 🏗️ 3 Struktur Dasar Alur Logika Pemrograman

Dalam dunia pemrograman komputer, semua program rumit di dunia (mulai dari Instagram, game Mobile Legends, hingga kecerdasan buatan) sebenarnya hanya dibangun dari kombinasi **3 struktur logika dasar**:

### 1. Struktur Sekuensial (Runtutan / Langkah Lurus)
Setiap langkah dijalankan berurutan satu per satu dari atas ke bawah tanpa ada langkah yang dilewati.
```text
[Langkah 1] ➡️ [Langkah 2] ➡️ [Langkah 3]
```

### 2. Struktur Percabangan (Selection / Decision)
Program melakukan pengecekan kondisi untuk menentukan cabang jalan mana yang harus diambil.
```text
           [Kondisi?]
           /        \
       (Ya)          (Tidak)
       /                \
  [Jalan A]          [Jalan B]
```

### 3. Struktur Perulangan (Iteration / Looping)
Program mengulang serangkaian langkah yang sama berulang kali selama suatu kondisi masih bernilai Benar (*True*).
```text
       ┌──► [Lakukan Tindakan] ──┐
       │                         │
  (Kondisi Masih Ya?)            │
       ▲                         ▼
       └─────────────────────────┘
```

## 3. 🎯 Contoh Kasus & Penerapan: Dari Mudah ke Sulit

### 🟢 LEVEL 1: MUDAH (Alur Sekuensial / Lurus)

#### 📌 Kasus 1.1: Menghitung Luas dan Keliling Segitiga Siku-Siku
**Skenario:** Pengguna memasukkan alas, tinggi, dan sisi miring segitiga. Program menghitung luas (`0.5 * alas * tinggi`) dan keliling (`alas + tinggi + sisi miring`), lalu menampilkan hasilnya ke layar.

```mermaid
flowchart TD
    Start([MULAI]) --> In[/Input: alas, tinggi, sisi_miring/]
    In --> P1["luas = 0.5 * alas * tinggi"]
    P1 --> P2["keliling = alas + tinggi + sisi_miring"]
    P2 --> Out[/Tampilkan luas dan keliling/]
    Out --> Finish([SELESAI])
```

#### Diagram Teks ASCII (Alternatif Tampilan):
```text
     ┌──────────────┐
     │    MULAI     │ (Terminator)
     └──────┬───────┘
            │
            ▼
     ┌──────────────┐
    ╱  Input: alas, ╱
   ╱ tinggi, miring╱ (Input)
  └───────┬───────┘
          │
          ▼
   ┌──────────────┐
   │ Hitung Luas  │ (Process)
   │ Hitung Kel.  │
   └──────┬───────┘
          │
          ▼
     ┌──────────────┐
    ╱ Tampilkan Luas╱
   ╱  dan Keliling ╱ (Output)
  └───────┬───────┘
          │
          ▼
     ┌──────────────┐
     │   SELESAI    │ (Terminator)
     └──────────────┘
```


````python
# Penerapan Kasus 1.1: Menghitung Luas & Keliling Segitiga Siku-Siku
print("=== PROGRAM HITUNG SEGITIGA SIKU-SIKU ===")

# 1. INPUT (Jajar Genjang)
alas = 6         # cm
tinggi = 8       # cm
sisi_miring = 10 # cm

# 2. PROSES (Persegi Panjang)
luas = 0.5 * alas * tinggi
keliling = alas + tinggi + sisi_miring

# 3. OUTPUT (Jajar Genjang)
print(f"Alas        : {alas} cm")
print(f"Tinggi      : {tinggi} cm")
print(f"Sisi Miring : {sisi_miring} cm")
print("-" * 35)
print(f"Luas        : {luas} cm²")
print(f"Keliling    : {keliling} cm")
````

#### 📌 Kasus 1.2: Konversi Suhu Celcius ke Reamur & Fahrenheit
**Skenario:** Memasukkan suhu dalam satuan Celcius, lalu mengonversinya ke Reamur (`4/5 * C`) dan Fahrenheit (`(9/5 * C) + 32`).

```mermaid
flowchart TD
    A([MULAI]) --> B[/Input: suhu_celcius/]
    B --> C["reamur = (4/5) * suhu_celcius"]
    C --> D["fahrenheit = (9/5 * suhu_celcius) + 32"]
    D --> E[/Tampilkan reamur dan fahrenheit/]
    E --> F([SELESAI])
```


````python
# Penerapan Kasus 1.2: Konversi Suhu
print("=== PROGRAM KONVERSI SUHU CELCIUS ===")

# Input
c = 30  # derajat Celcius

# Proses
r = (4 / 5) * c
f = (9 / 5 * c) + 32

# Output
print(f"Suhu Asal : {c}°C")
print(f"Ke Reamur : {r}°R")
print(f"Ke Fahr.  : {f}°F")
````

### 🟡 LEVEL 2: SEDANG (Alur Percabangan / Decision)

Pada level sedang, program tidak hanya berjalan lurus, melainkan harus **membuat keputusan** berdasarkan kondisi data masukan (*Branching*).

---
#### 📌 Kasus 2.1: Menentukan Kelulusan & Predikat Nilai Siswa
**Aturan Kasus:**
- Siswa memasukkan nama dan nilai ulangan (0 - 100).
- Jika `nilai >= 85` ➡️ Predikat **A (Sangat Memuaskan)**.
- Jika `nilai >= 75` dan `< 85` ➡️ Predikat **B (Lulus KKM)**.
- Jika `nilai < 75` ➡️ Predikat **C (Wajib Remedial)**.

```mermaid
flowchart TD
    Start([MULAI]) --> In[/Input: nama, nilai/]
    In --> Dec1{"Apakah nilai >= 85?"}
    
    Dec1 -- Ya --> P1["predikat = 'A (Sangat Memuaskan)'<br>status = 'LULUS'"]
    Dec1 -- Tidak --> Dec2{"Apakah nilai >= 75?"}
    
    Dec2 -- Ya --> P2["predikat = 'B (Lulus KKM)'<br>status = 'LULUS'"]
    Dec2 -- Tidak --> P3["predikat = 'C (Wajib Remedial)'<br>status = 'BELUM LULUS'"]
    
    P1 --> Out[/Tampilkan nama, nilai, status, predikat/]
    P2 --> Out
    P3 --> Out
    Out --> Finish([SELESAI])
```


````python
# Penerapan Kasus 2.1: Kelulusan & Predikat Nilai Siswa
nama_siswa = "Aulia"
nilai_siswa = 88

# Alur Percabangan Logika
if nilai_siswa >= 85:
    status = "LULUS 🎉"
    predikat = "A (Sangat Memuaskan)"
elif nilai_siswa >= 75:
    status = "LULUS 👍"
    predikat = "B (Lulus KKM)"
else:
    status = "BELUM LULUS ⚠️"
    predikat = "C (Harap Mengikuti Remedial)"

print("=== RAPOR DIGITAL SISWA ===")
print(f"Nama Siswa : {nama_siswa}")
print(f"Nilai      : {nilai_siswa}")
print(f"Status     : {status}")
print(f"Predikat   : {predikat}")
````

#### 📌 Kasus 2.2: Sistem Kasir Diskon Belanja Supermarket
**Aturan Kasus:**
- Pelanggan memasukkan total belanja.
- Jika total belanja **>= Rp 100.000**, pelanggan mendapat **diskon 10%**.
- Jika di bawah itu, diskon adalah **Rp 0**.
- Hitung total bayar akhir yang harus dibayarkan.

```mermaid
flowchart TD
    A([MULAI]) --> B[/Input: total_belanja/]
    B --> C{"total_belanja >= 100000?"}
    C -- Ya --> D["diskon = 0.10 * total_belanja"]
    C -- Tidak --> E["diskon = 0"]
    D --> F["total_bayar = total_belanja - diskon"]
    E --> F
    F --> G[/Tampilkan total_belanja, diskon, total_bayar/]
    G --> H([SELESAI])
```


````python
# Penerapan Kasus 2.2: Kasir Supermarket dengan Diskon
total_belanja = 125000  # Coba ubah menjadi 80000 untuk melihat kondisi kedua

# Logika Decision (Belah Ketupat)
if total_belanja >= 100000:
    diskon = 0.10 * total_belanja
    keterangan = "Selamat! Anda mendapatkan Diskon 10%"
else:
    diskon = 0
    keterangan = "Belanja belum mencapai Rp 100.000, belum dapat diskon"

total_bayar = total_belanja - diskon

print("=== STRUK PEMBAYARAN KASIR ===")
print(f"Total Belanja : Rp {total_belanja:,}")
print(f"Potongan      : Rp {int(diskon):,}")
print(f"Info          : {keterangan}")
print("-" * 35)
print(f"TOTAL BAYAR   : Rp {int(total_bayar):,}")
````

### 🔴 LEVEL 3: MENANTANG / SULIT (Alur Perulangan & Logika Kompleks)

Level menantang ini menggabungkan **Perulangan (Looping)**, **Percabangan (Decision)**, dan **Variabel Penghitung (Counter)**.

---
#### 📌 Kasus 3.1: Countdown Peluncuran Roket (Looping)
**Skenario:** Roket akan diluncurkan dengan hitungan mundur dari angka 5 ke 1. Setiap hitungan dicetak ke layar. Ketika hitungan mencapai 0, cetak kalimat `"ROKET MELUNCUR KE LUAR ANGKASA! 🚀"`.

```mermaid
flowchart TD
    Start([MULAI]) --> Init["Preparation: hitungan = 5"]
    Init --> Check{"Apakah hitungan > 0?"}
    
    Check -- Ya --> PrintNum[/Tampilkan: hitungan/]
    PrintNum --> Decr["hitungan = hitungan - 1"]
    Decr --> Check
    
    Check -- Tidak --> Blast[/Tampilkan: ROKET MELUNCUR! 🚀/]
    Blast --> Finish([SELESAI])
```


````python
# Penerapan Kasus 3.1: Countdown Peluncuran Roket
import time

print("=== PERSIAPAN PELUNCURAN ROKET PUSAT ANTARIKSA ===")
hitungan = 5

while hitungan > 0:
    print(f"Hitung mundur: {hitungan} detik...")
    hitungan = hitungan - 1
    # time.sleep(1)  # Simulasi jeda 1 detik

print("🔥 3... 2... 1... ROKET BERHASIL MELUNCUR KE LUAR ANGKASA! 🚀🌌")
````

#### 📌 Kasus 3.2: Mesin Verifikasi PIN ATM (Batas 3 Kali Percobaan)
**Skenario Dunia Nyata:**
- PIN yang benar tersimpan di sistem adalah `"1234"`.
- Pengguna diberi kesempatan mencoba maksimal **3 kali**.
- Jika PIN benar:
  - Tampilkan pesan `"PIN Benar! Silakan pilih transaksi"` dan program selesai.
- Jika PIN salah:
  - Kesempatan berkurang 1.
  - Jika masih ada kesempatan, pengguna diminta memasukkan PIN lagi (*Looping*).
  - Jika kesempatan sudah habis (3 kali salah), kartu **DIBLOKIR demi keamanan**!

```mermaid
flowchart TD
    Start([MULAI]) --> Init["PIN_BENAR = '1234'<br>kesempatan = 3<br>berhasil = False"]
    Init --> Check{"kesempatan > 0 DAN berhasil == False?"}
    
    Check -- Ya --> InputPIN[/Input: input_pin/]
    InputPIN --> Verify{"Apakah input_pin == PIN_BENAR?"}
    
    Verify -- Ya --> Succ["berhasil = True"]
    Succ --> MsgSucc[/Tampilkan: PIN Benar! Akses Diterima/]
    MsgSucc --> Check
    
    Verify -- Tidak --> Fail["kesempatan = kesempatan - 1"]
    Fail --> MsgFail[/Tampilkan: PIN Salah! Sisa Kesempatan/]
    MsgFail --> Check
    
    Check -- Tidak --> FinalCheck{"Apakah berhasil == False?"}
    FinalCheck -- Ya --> Block[/Tampilkan: KARTU ANDA DIBLOKIR! ❌/]
    FinalCheck -- Tidak --> EndLoop[Lanjut ke Menu Transaksi]
    
    Block --> Finish([SELESAI])
    EndLoop --> Finish
```


````python
# Penerapan Kasus 3.2: Simulasi Verifikasi PIN ATM
PIN_BENAR = "1234"
kesempatan = 3
berhasil = False

# Daftar percobaan (simulasi input dari pengguna)
simulasi_input = ["9999", "0000", "1234"] 
indeks_input = 0

print("=== MESIN ATM BANK GG ===")

while kesempatan > 0 and not berhasil:
    # Mengambil input PIN simulasi
    pin_masuk = simulasi_input[indeks_input]
    indeks_input += 1
    
    print(f"\n[Pengguna mengetik PIN: {pin_masuk}]")
    
    if pin_masuk == PIN_BENAR:
        berhasil = True
        print("✅ PIN BENAR! Selamat datang, akses rekening terbuka.")
    else:
        kesempatan -= 1
        print(f"❌ PIN SALAH! Sisa kesempatan Anda: {kesempatan} kali.")

# Keputusan akhir setelah perulangan selesai
if not berhasil:
    print("\n🚨 PERINGATAN: Anda salah memasukkan PIN 3 kali berturut-turut.")
    print("🔒 KARTU ATM ANDA DIBLOKIR DEMI KEAMANAN! Silakan hubungi Customer Service.")
else:
    print("\n💵 Silakan pilih transaksi: [1] Tarik Tunai  [2] Cek Saldo  [3] Transfer")
````

### 📝 Latihan 1: Analisis Jalur Nilai (Tracing Flowchart)
Perhatikan alur logika di bawah ini:
```text
[MULAI]
   │
   ▼
[angka = 14]
   │
   ▼
{ Apakah angka % 2 == 0? }
   │                  │
 (Ya)               (Tidak)
   │                  │
   ▼                  ▼
[jenis = 'GENAP']   [jenis = 'GANJIL']
   │                  │
   └────────┬─────────┘
            ▼
    [/ Tampilkan jenis /]
            │
            ▼
        [SELESAI]
```

**Pertanyaan:**
1. Apa output yang dicetak ke layar jika nilai `angka = 14`?
2. Apa output yang dicetak ke layar jika nilai `angka = 27`?
3. Simbol apa yang digunakan pada baris `{ Apakah angka % 2 == 0? }`?


````python
# Cek Jawaban Latihan 1 dengan Python:
def cek_ganjil_genap(angka):
    if angka % 2 == 0:
        return "GENAP"
    else:
        return "GANJIL"

print("Jawaban untuk angka 14:", cek_ganjil_genap(14))
print("Jawaban untuk angka 27:", cek_ganjil_genap(27))
print("Simbol yang dipakai    : Decision (Belah Ketupat / Diamond)")
````

  2. Jika **Hujan Lebat** ➡️ Olahraga di dalam gedung / aula sekolah (pakai seragam kaos + jaket).
  3. Jika **Cerah / Berawan** ➡️ Olahraga di lapangan terbuka (pakai seragam olahraga biasa + bawa botol air minum).

### 🛠️ Panduan 4 Langkah Merancang Flowchart Sendiri:
1. **Identifikasi Masalah:** Apa tujuan utama program ini?
2. **Tentukan I-P-O:**
   - **Input:** Data apa saja yang dimasukkan oleh user?
   - **Proses:** Rumus apa yang dihitung? Keputusan apa yang diambil?
   - **Output:** Apa hasil akhir yang ditampilkan?
3. **Pilih Simbol yang Tepat:**
   - Awal & Akhir wajib `Oval (Terminator)`.
   - Masukan & Tampilan pakai `Jajar Genjang (I/O)`.
   - Tindakan & Rumus pakai `Persegi Panjang (Process)`.
   - Pertanyaan kondisi pakai `Belah Ketupat (Decision)`.
4. **Periksa Jalur:** Pastikan tidak ada panah buntu yang tidak menuju ke simbol `Selesai`!
